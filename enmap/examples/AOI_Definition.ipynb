```python
import json
from datetime import datetime
from shapely.geometry import Polygon
from ipyleaflet import Map, DrawControl, basemap_to_tiles, basemaps
```

```python
# Define the datetime range here
START_DATETIME = "2024-09-10"
END_DATETIME = "2024-09-11"
```

```python
aoi_geometry = None

m = Map(center=(38.8, 23.5), zoom=10)

m.add_layer(basemap_to_tiles(basemaps.Esri.WorldImagery))

draw_control = DrawControl(
    polygon={"shapeOptions": {"color": "#00ff00"}},
    rectangle={"shapeOptions": {"color": "#0000FF"}},
    circle={},
    circlemarker={},
    marker={},
    polyline={}
)

def handle_draw(target, action, geo_json):
    global aoi_geometry

    if action == "created":
        aoi_geometry = geo_json["geometry"]

        print("Polygon stored.")
        print(aoi_geometry)

draw_control.on_draw(handle_draw)

m.add_control(draw_control)

m
```

```python
if aoi_geometry is None:
    raise RuntimeError("Draw a polygon first.")

query = {
    "datetime": f"{START_DATETIME}/{END_DATETIME}",
    "aoi": aoi_geometry
}

with open("user_input_example.json", "w") as f:
    json.dump(query, f, indent=4)

print("user_input_example.json saved.")
```
